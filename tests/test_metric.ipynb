```python
# Method 1: Using recursion
def flatten_recursive(nested_list):
    flat_list = []
    for item in nested_list:
        if isinstance(item, list):
            flat_list.extend(flatten_recursive(item))
        else:
            flat_list.append(item)
    return flat_list

# Method 2: Using a list comprehension with recursion
def flatten_comprehension(nested_list):
    return [item for sublist in nested_list for item in (flatten_comprehension(sublist) if isinstance(sublist, list) else [sublist])]

# Method 3: Using itertools.chain
from itertools import chain
def flatten_itertools(nested_list):
    return list(chain(*nested_list))

# Example usage
nested = [[1, 2, [3, 4]], [5, 6], 7]

# Method 1 result
print(flatten_recursive(nested))  # Output: [1, 2, 3, 4, 5, 6, 7]

# Method 2 result
print(flatten_comprehension(nested))  # Output: [1, 2, 3, 4, 5, 6, 7]

# Method 3 result (only works for one level of nesting)
print(flatten_itertools([[1, 2], [3, 4]]))  # Output: [1, 2, 3, 4]
```
