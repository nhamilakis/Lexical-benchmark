```python
import matplotlib.pyplot as plt
import numpy as np

class PlotSettings:
    COLORS = {
        'human_CDI': '#d62728',
        'CHILDES': 'orange',
        '0.3': '#DCE3E8',
        '0.6': '#8A9AA0',
        '0.8': '#3F5C6A',
        '1.0': '#163B4A',
        '1.5': '#021518',
        'LSTM': '#ff9896',
        'Transformer': '#c49c94'
    }
    
    LINESTYLES = {
        'CHILDES': '-',
        'child': '-',
        '0.3': '-',
        '0.6': '-.',
        '0.8': '--',
        '1.0': ':',
        '1.5': '-'
    }
    
    FIG_SIZE = (10, 10)
    
    @classmethod
    def configure_plot(cls, title=None, model=None):
        plt.figure(figsize=cls.FIG_SIZE)
        
        if title or model:
            plt.title(title or model, fontsize=18, fontweight='bold')
        
        plt.xticks(fontsize=18)
        plt.yticks(fontsize=18)
        plt.grid(True, linestyle=':', linewidth=1, color='#bbbbbb')
    
    @classmethod
    def configure_legend(cls):
        plt.legend(
            title='Settings', 
            title_fontsize=28, 
            fontsize=28, 
            loc='upper left', 
            bbox_to_anchor=(1.1, 1)
        )
    
    @classmethod
    def save_figure(cls, filepath):
        plt.tight_layout()
        plt.savefig(filepath, bbox_inches='tight')
        plt.close()

```

```python
# Example usage
def plot_comparison():
    # Prepare data
    x = np.linspace(0, 10, 100)
    
    # Create plot
    PlotSettings.configure_plot(model='Performance Comparison')
    
    # Plot multiple lines with consistent styling
    plt.plot(x, np.sin(x), 
             color=PlotSettings.COLORS['CHILDES'], 
             linestyle=PlotSettings.LINESTYLES['CHILDES'], 
             label='CHILDES')
    
    plt.plot(x, np.cos(x), 
             color=PlotSettings.COLORS['LSTM'], 
             linestyle=PlotSettings.LINESTYLES['0.6'], 
             label='LSTM')
    
    # Configure legend
    PlotSettings.configure_legend()
    
    # Save figure
    PlotSettings.save_figure('performance_plot.png')

# Run the example
plot_comparison()
```

```python


def write_to_word_dict(
    dataset: str,
    chunk: str | int,
    model_type: str,
    temp: str | float,
    word_count: int,
    words_dict: dict[str, dict]
) -> dict[str, dict]:
    """Write word count to a nested dictionary structure.
    
    Args:
        dataset: Name of the dataset
        chunk: Chunk identifier (string or integer)
        model_type: Type of model used
        temp: Temperature value
        word_count: Number of words to store
        words_dict: Existing dictionary to update
        
    Returns:
        Updated dictionary with new word count
    """
    # Create nested structure if it doesn't exist
    if dataset not in words_dict:
        words_dict[dataset] = {}
    
    chunk_str = str(chunk)
    if chunk_str not in words_dict[dataset]:
        words_dict[dataset][chunk_str] = {}
        
    if model_type not in words_dict[dataset][chunk_str]:
        words_dict[dataset][chunk_str][model_type] = {}
        
    temp_str = str(temp)
    words_dict[dataset][chunk_str][model_type][temp_str] = word_count
    
    return words_dict


def test_write_to_dict() -> None:
    """Test the write_to_word_dict function."""
    # Initialize empty dictionary
    test_dict: dict[str, dict] = {}
    
    # Test cases
    test_cases = [
        ("dataset1", 1, "gpt2", 0.7, 100),
        ("dataset1", "2", "gpt3", "0.8", 200),
        ("dataset2", 1, "gpt2", 0.9, 300)
    ]
    
    # Run test cases
    for dataset, chunk, model, temp, count in test_cases:
        test_dict = write_to_word_dict(
            dataset, chunk, model, temp, count, test_dict
        )
        
        # Verify storage
        try:
            stored_count = test_dict[dataset][str(chunk)][model][str(temp)]
            assert stored_count == count
            print(f"Successfully stored {count} words for {dataset}")
        except (KeyError, AssertionError) as e:
            print(f"Error in storage: {e}")

if __name__ == "__main__":
    test_write_to_dict()
```
